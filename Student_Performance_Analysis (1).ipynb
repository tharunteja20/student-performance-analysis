```python
import pandas as pd
df = pd.read_csv('StudentsPerformance.csv')
df.head()
```

```python
df.groupby('test preparation course')['math score'].mean()
```

```python
df.groupby('parental level of education')['math score'].mean().sort_values()
```

```python
import matplotlib.pyplot as plt
df.groupby('parental level of education')['math score'].mean().sort_values().plot(kind='bar')
plt.title('Average Math Score by Parental Level of Education')
plt.ylabel('Math Score')
plt.show()
```

```python
df['math score'].hist(bins=20)
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Number of Students')
plt.show()
```

This project analyzes exam performance for 1000 students using Python and Pandas. The dataset had no missing values or duplicates. Students who completed a test preparation course scored an average of 69.7 in math, about 5.6 points higher than those who didn't. Math scores also rose steadily with parental education level, from 62.1 (high school) to 69.7 (master's degree) — a 7.6 point spread. Math scores overall follow a roughly normal distribution centered in the mid-to-high 60s.
